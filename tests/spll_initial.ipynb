```python
import numpy as np
dc_currents = np.logspace(-12,-6 ,5)
spike_freq = []
for idx,dc_current in enumerate(dc_currents):

    chip.update_parameter_in_dict(name='input_osc_PI',value=dc_current)
    bitstream = chip.create_fifo(type='slow')
    chip.program_fifo(bitstream,fifotype='fast')
    # chip.send_spiketrain(['IN' for i in range(100)],[2000*i for i in range(100)],width=300)
    chip.start_experiment()
    time.sleep(3)
    received = chip.read_gpios()
    spike_freq.append(np.mean(np.diff(received[1])))
    print(idx,dc_current,np.mean(np.diff(received[1])))
    chip.stop_experiment()
```
